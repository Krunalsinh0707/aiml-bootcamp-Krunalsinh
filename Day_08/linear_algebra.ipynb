TASK 04
Prove that dot product zero means perpendicular

Generate many pairs of 2D vectors. For each, compute both the dot product and the angle.
• Plot dot product against angle. Where does the dot product cross zero?

```python
import numpy as np
import matplotlib.pyplot as plt


angles = []
dots = []

base = np.array([1.0, 0.0])      # Fixed vector

for deg in range(0, 181, 5):

    rad = np.radians(deg)

    v = np.array([
        np.cos(rad),
        np.sin(rad)
    ])

    angles.append(deg)
    dots.append(base @ v)

plt.figure(figsize=(8,5))

plt.plot(angles, dots, marker="o")

plt.axhline(0, color="red", linestyle="--", label="Dot Product = 0")
plt.axvline(90, color="green", linestyle="--", label="90°")

plt.xlabel("Angle (degrees)")
plt.ylabel("Dot Product")
plt.title("Dot Product vs Angle")

plt.grid(True)
plt.legend()

plt.show()
```

Confirm: every pair with a 90° angle has a dot product of (near) zero, and vice versa

```python
print("Angle    Dot Product")
print("-"*25)

for a, d in zip(angles, dots):
    if abs(a-90) <= 5 or abs(d) < 0.05:
        print(f"{a:5}°    {d:.5f}")

```

TASK 06
Detect redundant features

Build a dataset with 5 features where two are linearly dependent — e.g. 'height_cm' and 'height_inches' (one is a
fixed multiple of the other).

• Compute the rank of the feature matrix. It should be less than 5. Explain what that tells you.

• Identify which features are redundant. (Hint: check correlations, or drop features one at a time and watch the
rank.)

• Remove the redundant feature and confirm the rank is now full.

• Write two sentences on why perfectly correlated features cause problems for linear models specifically.

```python
import numpy as np

# Random number generator
rng = np.random.default_rng(0)

# Create 5 features
height_cm = rng.normal(170, 10, 100)
height_in = height_cm / 2.54          # Perfectly dependent
weight = rng.normal(70, 15, 100)
age = rng.integers(18, 60, 100)
salary = rng.normal(8, 2, 100)

# Feature names
feature_names = [
    "height_cm",
    "height_in",
    "weight",
    "age",
    "salary"
]

# Stack into feature matrix
X = np.column_stack((height_cm, height_in, weight, age, salary))

# Matrix Rank

rank = np.linalg.matrix_rank(X)

print("Original Matrix Shape:", X.shape)
print("Original Rank:", rank)

# Correlation Matrix

print("\nCorrelation Matrix:")
corr = np.corrcoef(X, rowvar=False)
print(np.round(corr, 3))

# Find redundant feature


print("\nRank after removing each feature:")

for i, feature in enumerate(feature_names):
    X_new = np.delete(X, i, axis=1)
    new_rank = np.linalg.matrix_rank(X_new)
    print(f"Remove {feature:10s} -> Rank = {new_rank}")

# Remove redundant feature
X_reduced = np.delete(X, 1, axis=1)   # Remove height_in

print("\nReduced Matrix Shape:", X_reduced.shape)
print("New Rank:", np.linalg.matrix_rank(X_reduced))
```

TASK 03
Implement Gram-Schmidt orthogonalisation

Implement gram_schmidt(vectors) that returns an orthonormal basis. The recipe: for each vector, subtract
its projection onto every basis vector already found, then normalise.

• Run it on three independent 3D vectors.

• Prove it worked: compute Q @ Q.T and confirm it is the identity matrix. Explain why that proves orthonormality.

• Check that the result spans the same space as the input (same rank).

• Now feed it a dependent set — make the third vector a sum of the first two. What does your function do? Make
sure it does not crash or produce a garbage near-zero vector.

• Verify against NumPy: np.linalg.qr does the same thing. Compare your basis to its Q. (They may differ by
sign — why is that fine?)

```python
import numpy as np


def gram_schmidt(vectors, tol=1e-10):
    basis = []

    for v in vectors:
        v = v.astype(float).copy()

        # Remove projections onto existing basis vectors
        for q in basis:
            projection = np.dot(v, q) * q
            v = v - projection

        # Compute length of the remaining vector
        norm = np.linalg.norm(v)

        # Skip dependent (near-zero) vectors
        if norm > tol:
            basis.append(v / norm)

    return np.array(basis)

# Independent vectors

vectors = np.array([
    [1, 1, 0],
    [1, 0, 1],
    [0, 1, 1]
], dtype=float)

Q = gram_schmidt(vectors)

print("Orthonormal Basis (Q):")
print(Q)

# Verify orthonormality

print("\nQ @ Q.T")
print(Q @ Q.T)

print("\nIs it close to Identity?")
print(np.allclose(Q @ Q.T, np.eye(len(Q))))

# Compare ranks

print("\nOriginal Rank :", np.linalg.matrix_rank(vectors))
print("Basis Rank    :", np.linalg.matrix_rank(Q))


# Dependent vectors
# Third vector = first + second

dependent_vectors = np.array([
    [1, 1, 0],
    [1, 0, 1],
    [2, 1, 1]
], dtype=float)

Q_dep = gram_schmidt(dependent_vectors)

print("\nDependent Set Basis:")
print(Q_dep)

print("\nRank of dependent set:")
print(np.linalg.matrix_rank(dependent_vectors))

print("Basis Rank:")
print(np.linalg.matrix_rank(Q_dep))


# Compare with NumPy QR
Q_numpy, R = np.linalg.qr(vectors.T)

print("\nNumPy QR Basis:")
print(Q_numpy.T)

print("\nOur Basis:")
print(Q)

print("\nAbsolute difference (sign ignored):")
print(np.allclose(np.abs(Q), np.abs(Q_numpy.T)))
```
