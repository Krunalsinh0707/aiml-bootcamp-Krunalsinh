Importing Libraries and Load The Data

```python
import pandas as pd 
import numpy as np 
import seaborn as sns 
import matplotlib.pyplot as plt 
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score, precision_score, recall_score, f1_score
from sklearn.metrics import roc_curve, roc_auc_score, precision_recall_curve

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (8, 4.5)

df = pd.read_csv("titanic.csv")
print(f"Loaded: {df.shape[0]} passenger records, {df.shape[1]} columns")
df.head()
```

TASK 01 · Encode, split, and standardize

• Encode binary categoricals: convert Sex to binary (female = 1, male = 0).

```python
df_clean = df.copy()
df_clean['Sex'] = (df_clean['Sex'] == 'female').astype(int)

df_clean.head()
```

• Separate features X and binary target y = Survived.

```python
x = df_clean.drop(columns=['Survived'])
y = df_clean['Survived'].values

print(x.shape)
print(y.shape)
```

• Split into training (80%) and test (20%) sets with stratify=y and fixed random_state=42.

```python
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.20, random_state=42, stratify=y
)
```

• Standardize features using StandardScaler fit strictly on training data.

```python
scaler = StandardScaler().fit(x_train)
x_train_scaled = scaler.transform(x_train)
x_test_scaled = scaler.transform(x_test)

print(f"\nTraining set: {x_train_scaled.shape[0]} samples")
print(f"Testing set:  {x_test_scaled.shape[0]} samples")
```

TASK 02 · The Sigmoid function and Log-Loss

• Write a vectorized sigmoid function: sigmoid(z) = 1 / (1 + exp(-z)).

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-np.clip(z, -250, 250)))
```

• Plot the sigmoid S-curve from $z = -10$ to $z = +10$ and highlight the 0.5 decision threshold.

```python
z_vals = np.linspace(-10, 10, 300)
sig_vals = sigmoid(z_vals)

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(z_vals, sig_vals, color="#F05E22", linewidth=2.5, label="$\\sigma(z) = \\frac{1}{1 + e^{-z}}$")
ax.axhline(0.5, color="#1A1D1D", linestyle="--", linewidth=1, label="Threshold $\\sigma(z)=0.5$")
ax.axvline(0.0, color="#64748B", linestyle=":", linewidth=1)
ax.set_title("The Sigmoid Activation Function (Mapping Scores to Probabilities)", fontsize=13, pad=10)
ax.set_xlabel("Linear Score $z = w^T x + b$")
ax.set_ylabel("Predicted Probability $P(y=1)$")
ax.set_ylim(-0.05, 1.05)
ax.legend(frameon=True)
plt.tight_layout()
plt.show()
```

• Compute Binary Cross-Entropy penalty by hand for $y=1, p=0.9$ vs $y=1, p=0.05$.


```python
p = np.linspace(0.05, 0.9, 200)
loss_y1 = -np.log(p)       # When true y = 1
loss_y0 = -np.log(1 - p)   # When true y = 0

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(p, loss_y1, color="#F05E22", linewidth=2, label="Loss when True $y=1$ ($-\ln(p)$)")
ax.plot(p, loss_y0, color="#3B82F6", linewidth=2, linestyle="--", label="Loss when True $y=0$ ($-\ln(1-p)$)")
ax.set_title("Binary Cross-Entropy Loss Penalty Curves", fontsize=13, pad=10)
ax.set_xlabel("Predicted Probability $\hat{p}$")
ax.set_ylabel("Loss / Penalty")
ax.set_ylim(0, 5)
ax.legend(frameon=True)
plt.tight_layout()
plt.show()
```

• Implement class LogisticRegressionScratch with weights $w$ and bias $b$.

• Run gradient descent loop: gradients $\nabla_w J = \frac{1}{m} X^T (p - y)$ and $db = \frac{1}{m} \sum (p - y)$.

• Plot the training loss history over 2000 iterations to observe monotonic convergence.

```python
class LogisticRegressionScratch:
    def __init__(self, lr=0.1, n_iters=1500):
        self.lr = lr
        self.n_iters = n_iters
        self.w = None
        self.b = None
        self.loss_history = []

    def fit(self, X, y):
        m, n = X.shape
        self.w = np.zeros(n)
        self.b = 0.0
        self.loss_history = []

        for i in range(self.n_iters):
            # Forward pass: linear combination + sigmoid
            z = X @ self.w + self.b
            p = sigmoid(z)

            # Compute cross-entropy loss (with epsilon clipping to avoid log(0))
            p_clip = np.clip(p, 1e-15, 1 - 1e-15)
            loss = - (1/m) * np.sum(y * np.log(p_clip) + (1 - y) * np.log(1 - p_clip))
            self.loss_history.append(loss)

            # Gradient computation
            dz = p - y
            dw = (1/m) * (X.T @ dz)
            db = (1/m) * np.sum(dz)

            # Parameter updates
            self.w -= self.lr * dw
            self.b -= self.lr * db

    def predict_proba(self, X):
        z = X @ self.w + self.b
        return sigmoid(z)

    def predict(self, X, threshold=0.5):
        return (self.predict_proba(X) >= threshold).astype(int)

# Train from scratch
scratch_model = LogisticRegressionScratch(lr=0.15, n_iters=2000)
scratch_model.fit(x_train_scaled, y_train)

print("From-Scratch Training Complete!")
print(f"Learned Intercept (bias b): {scratch_model.b:.4f}")
print("Learned Weights (w):")
for feat, weight in zip(x.columns, scratch_model.w):
    print(f"  {feat:10s}: {weight:.4f}")
```

TASK 04 · Method 2 — Scikit-Learn parity proof

• Fit LogisticRegression(penalty=None, solver='lbfgs') on the scaled training set.

• Display learned intercept and feature weights side-by-side with your from-scratch model.

• Show numerically that weights match within $10^{-3}$ and predictions match on the test set.

```python

# Fit unregularized logistic regression for direct parity comparison
sk_model = LogisticRegression(penalty=None, solver='lbfgs', max_iter=1000)
sk_model.fit(x_train_scaled, y_train)

# Compare learned parameters side-by-side
parity_df = pd.DataFrame({
    'Feature': ['Intercept (bias)'] + list(x.columns),
    'From-Scratch (NumPy)': [scratch_model.b] + list(scratch_model.w),
    'Scikit-Learn': [sk_model.intercept_[0]] + list(sk_model.coef_[0]),
})
parity_df['Abs Difference'] = np.abs(parity_df['From-Scratch (NumPy)'] - parity_df['Scikit-Learn'])

print("Parameter Parity Comparison:")
print(parity_df.to_string(index=False))

# Predict on test set
pred_scratch = scratch_model.predict(x_test_scaled)
pred_sklearn = sk_model.predict(x_test_scaled)
print(f"\nPredictions Match Rate: {(pred_scratch == pred_sklearn).mean()*100:.1f}%")
```

TASK 06 · Decision Threshold Tuning and the ROC-AUC Curve

• Extract predicted probabilities: probs = model.predict_proba(X_te)[:, 1].

```python
test_probs = sk_model.predict_proba(x_test_scaled)[:, 1]
```

• Sweep threshold across $[0.2, 0.5, 0.8]$ and document how Precision and Recall trade off.

```python
thresholds = [0.2, 0.5, 0.8]

for threshold in thresholds:
    y_pred = (test_probs >= threshold).astype(int)

    precision = precision_score(y_test, y_pred)
    recall = recall_score(y_test, y_pred)

    print(f"Threshold: {threshold}")
    print(f"Precision: {precision:.3f}")
    print(f"Recall:    {recall:.3f}")
```

```python
fpr, tpr, thresholds = roc_curve(y_test, test_probs)
auc_score = roc_auc_score(y_test, test_probs)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 4.5))

# Plot 1: ROC Curve
ax1.plot(fpr, tpr, color="#F05E22", linewidth=2.5, label=f"Logistic Regression (AUC = {auc_score:.3f})")
ax1.plot([0, 1], [0, 1], color="#1A1D1D", linestyle="--", linewidth=1, label="Random Guess (AUC = 0.500)")
ax1.set_title("Receiver Operating Characteristic (ROC) Curve", fontsize=12)
ax1.set_xlabel("False Positive Rate (1 - Specificity)")
ax1.set_ylabel("True Positive Rate (Recall)")
ax1.legend(loc="lower right")

# Plot 2: Precision vs Recall vs Threshold
precisions, recalls, pr_thresholds = precision_recall_curve(y_test, test_probs)
ax2.plot(pr_thresholds, precisions[:-1], color="#3B82F6", linewidth=2, label="Precision")
ax2.plot(pr_thresholds, recalls[:-1], color="#F05E22", linewidth=2, label="Recall")
ax2.axvline(0.5, color="#1A1D1D", linestyle=":", label="Default Threshold (0.5)")
ax2.set_title("Precision & Recall vs Decision Threshold", fontsize=12)
ax2.set_xlabel("Classification Threshold")
ax2.set_ylabel("Score")
ax2.legend(loc="lower left")

plt.tight_layout()
plt.show()

print(f"ROC-AUC Score on Test Set: {auc_score:.4f}")
```

• Calculate the Odds Ratio for each feature: $\text{Odds Ratio} = e^{w_j}$.

```python
coef_df = pd.DataFrame({
    'Feature': x.columns,
    'Weight': sk_model.coef_[0],
    'Odds_Ratio': np.exp(sk_model.coef_[0])
}).sort_values(by='Weight', ascending=True)

print("Feature Weights & Odds Ratios:")
print(coef_df.sort_values(by='Weight', ascending=False).to_string(index=False))
```

• Generate a horizontal bar chart of feature weights with orange for positive and blue for negative impact.

```python
fig, ax = plt.subplots(figsize=(8, 4))
bar_colors = ["#F05E22" if w > 0 else "#3B82F6" for w in coef_df['Weight']]
ax.barh(coef_df['Feature'], coef_df['Weight'], color=bar_colors)
ax.axvline(0, color="#1A1D1D", linewidth=0.8)
ax.set_title("Logistic Regression Feature Weights (Orange = Promotes Survival, Blue = Reduces Survival)", fontsize=11, pad=10)
ax.set_xlabel("Weight (on standardized features)")
plt.tight_layout()
plt.show()
```
