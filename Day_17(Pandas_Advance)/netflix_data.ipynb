```python
import pandas as pd 


data = pd.read_csv("netflix_titles-selected-columns.csv")
print(data.shape)
data.info()
data.head()
```

```python
cleaned_data = data.dropna()
cleaned_data.info()
cleaned_data.head()
```

```python
# data['date_added'] = pd.to_datetime(data['date_added'].str.strip(),   format="%B %d, %Y")
data["release_year"] = pd.to_numeric(
    data["release_year"],
    errors="coerce"
)
data
```

```python
text_cols = data.select_dtypes(include="str").columns

data[text_cols] = data[text_cols].apply(
    lambda x: x.str.strip().str.title()
)
```
