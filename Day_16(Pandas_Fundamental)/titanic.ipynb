TASK 02 Three strategies for missing age

• Confirm how many age values are missing, and what fraction of the data that is.

```python
print("Missing age :" , missing_age)

print("Rate of Missing age :", missing_age/len(df))
```

• Strategy 1 — drop: remove rows with missing age. How many rows remain? What did the mean age become?

```python
original_count = df["age"].notnull().sum()
original_mean = df["age"].mean()
original_std = df["age"].std()
print(f"BEFORE: {original_count} known ages, mean={original_mean:.1f}, std={original_std:.1f}\n")

dropped = df.dropna(subset=["age"])
print(f"1. DROP:  {len(dropped)} rows remain (lost {len(df)-len(dropped)})")
print(f"          mean={dropped['age'].mean():.1f}, std={dropped['age'].std():.1f}")


print("Female Survival Rate  :",dropped[dropped["sex"] == "female"]["survived"].mean())
print("Male Survival Rate ",dropped[dropped["sex"] == "male"]["survived"].mean())
```

• Strategy 2 — mean fill: fill missing ages with the overall mean age. Check what happens to the standard
deviation, and explain why it changes.

```python
filled_mean = df.copy()
filled_mean["age"] = filled_mean["age"].fillna(df["age"].mean())
print(f"2. MEAN fill: all {len(filled_mean)} rows kept")
print(f"              mean={filled_mean['age'].mean():.1f}, std={filled_mean['age'].std():.1f}")
print("   note: std SHRANK -- filling with one value makes the data artificially less spread out")


print("Female Survival Rate  :",filled_mean[filled_mean["sex"] == "female"]["survived"].mean())
print("Male Survival Rate ",filled_mean[filled_mean["sex"] == "male"]["survived"].mean())
```

• Strategy 3 — group fill: fill each missing age with the median age of passengers in the same class and sex.
Why is this smarter than the overall mean?

```python
filled_group = df.copy()
filled_group["age"] = filled_group.groupby(["pclass", "sex"])["age"].transform(
    lambda x: x.fillna(x.median())
)
print(f"3. GROUP median fill: all {len(filled_group)} rows kept")
print(f"                      mean={filled_group['age'].mean():.1f}, std={filled_group['age'].std():.1f}")
print("   smarter: each missing age filled from similar passengers, spread better preserved")

print("Female Survival Rate  :",filled_group[filled_group["sex"] == "female"]["survived"].mean())
print("Male Survival Rate ",filled_group[filled_group["sex"] == "male"]["survived"].mean())
```

• State which one you would choose for this dataset, and justify it — think about how much data you'd lose,
whether the spread is preserved, and whether it could bias later analysis.

• Note explicitly: is there a single 'correct' answer, or does it depend on the goal? Give an example of a goal

Starter hints


### Missing-Age Strategies Comparison

1. **Dropping missing rows:** This is simple and keeps only the original age values, but it can cause a significant loss of data, especially when many ages are missing.


2. **Mean imputation:** Missing ages are replaced with the average age. It keeps all rows but can reduce the natural spread of the data and can be affected by extreme values.


3. **Median imputation:** Missing ages are replaced with the median age. It keeps all the data and is more robust to outliers, so it is generally a better choice for the Titanic age data.


4. **My choice:** I would choose **median imputation** for this dataset because it avoids losing passengers and preserves the data better than mean imputation when the age distribution is not perfectly balanced.


5. **No single correct answer:** There is no universally correct strategy; the choice depends on the goal. For example, if only a few rows have missing values, dropping those rows may be preferable to introducing estimated values.


6. **Trust and transparency:** If someone imputed 20% of a dataset without mentioning it, that could be a problem because a large portion of the data would contain estimated values. This could affect later analysis and make the results less trustworthy if readers do not know how the missing values were handled.

• Rank passengers by fare and find what fraction of the top-10% fare-payers survived, versus the bottom-10%.

```python
n = int(len(df) * 0.10)
print(n)

sorted_fair = df.sort_values("fare", ascending=False)

top_10 = sorted_fair.head(n)

bottom_10 = sorted_fair.tail(n)


top_survival = top_10["survived"].mean()

bottom_survival = bottom_10["survived"].mean()

print ("survival rate of the top fare payers " , top_survival)

print("Survival rate of Least  fare payer  " , bottom_survival)
```
